```python
import pandas as pd
import json
```

```python
new_df = pd.read_csv("data/new.csv", sep="\t", encoding="utf-8")
```

```python
new_df.to_csv(
    "data/data.tsv", sep="\t", mode="a", header=False, index=False, encoding="utf-8"
)
```

```python
with open("data/data.arff", "a", encoding="utf-8") as f:
    for _, row in new_df.iterrows():
        row = row.to_dict()
        for key in ["id", "Возраст", "Рост, см", "Вес, кг"]:
            if pd.isna(row[key]):
                row[key] = "?"
            else:
                row[key] = str(row[key])

        for key in [
            "Имя",
            "Пол",
            "Город",
            "Знакомлюсь для",
            "Семейное положение",
            "Дети",
            "Курение",
            "Алкоголь",
            "Доход",
            "Проживание",
            "Наличие автомобиля",
            "Знание языков",
            "Тело",
            "Цвет волос",
            "Цвет глаз",
            "Обо мне",
            "Кого ищу",
        ]:
            if pd.isna(row[key]):
                row[key] = "?"
            else:
                row[key] = f"'{row[key].replace("'", "\\'")}'"

        f.write(f"{','.join(row.values())}\n")
```

```python
with open("data/data.json", "r", encoding="utf-8") as f:
    df = json.load(f)

for _, row in new_df.iterrows():
    row = row.to_dict()
    for key in ["Рост, см", "Вес, кг"]:
        row[key] = int(row[key]) if pd.notna(row[key]) else None

    for key in ["Знакомлюсь для", "Знание языков"]:
        row[key] = row[key].split(", ") if pd.notna(row[key]) else []

    for key in [
        "Имя",
        "Пол",
        "Город",
        "Семейное положение",
        "Дети",
        "Курение",
        "Алкоголь",
        "Доход",
        "Проживание",
        "Наличие автомобиля",
        "Тело",
        "Цвет волос",
        "Цвет глаз",
        "Обо мне",
        "Кого ищу",
    ]:
        if pd.isna(row[key]):
            row[key] = None

    df["data"].append(row)

with open("data/data.json", "w", encoding="utf-8") as f:
    json.dump(df, f, ensure_ascii=False, indent=4)
```
